```python

import pandas as pd
import numpy as np

df = pd.read_csv('hacker_news.csv')

print(df.head())
print(df.tail())
```

```python
print(df.columns)
s = df['title']
print(s.head())
print(df.shape)

count_of_python = df['title'].str.contains('Python', case=False).sum()
print(f'Number of posts with Python in the title: {count_of_python}')
count_of_java = df['title'].str.contains('Java', case=False).sum()
print(f'Number of posts with Java in the title: {count_of_java}')



    
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import datetime as dt

df['created_at'] = pd.to_datetime(df['created_at'], format='%m/%d/%Y %H:%M')

sns.set_style('darkgrid')
plt.hist(df['created_at'], bins=50, align='left')  # use actual datetimes so the x-axis shows a timeline
plt.xlabel('Months')
plt.ylabel('Number of Posts')
plt.title('Distribution of Posts by Month')
plt.show()
```
